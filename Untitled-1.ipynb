```python
import sqlite3
import time
from urllib.parse import urljoin

import requests
from bs4 import BeautifulSoup
```

```python
START_URL = "https://phyzikyh.github.io/databuilding/store/books/page1.html"
API_URL = "https://phyzikyh.github.io/databuilding/store/api/books.json"
```

```python
conn = sqlite3.connect("collect.db")
cur = conn.cursor()
cur.execute("DROP TABLE IF EXISTS books")
cur.execute("""CREATE TABLE books(
    book_id TEXT PRIMARY KEY,
    title TEXT NOT NULL,
    author TEXT,
    category TEXT,
    price INTEGER,
    rating REAL,
    cover_url TEXT,
    source_url TEXT
)""")
conn.commit()
print("준비 완료")
```

```python
response = requests.get(START_URL, timeout=10)
response.raise_for_status()

print("상태 코드:", response.status_code)
print("응답 길이:", len(response.text))
```

```python
soup = BeautifulSoup(response.text, "html.parser")
books = soup.select("article.book")

print("현재 페이지 상품 수:", len(books))
```

```python
def parse_book(book, source_url):
    price_text = book.select_one(".price").get_text(strip=True)

    return {
        "book_id": book["data-id"],
        "title": book.select_one(".title").get_text(strip=True),
        "author": book.select_one(".author").get_text(strip=True),
        "category": book["data-category-name"],
        "price": int(price_text.replace(",", "").replace("원", "")),
        "rating": float(book.select_one(".rating")["data-score"]),
        "cover_url": book.select_one("img.cover")["src"],
        "source_url": source_url,
    }

one = parse_book(book, START_URL)
print(one)
```

```python
page_rows = [
    parse_book(book, START_URL)
    for book in soup.select("article.book")
]

print("현재 페이지 수집:", len(page_rows), "건")
print(page_rows[0])
```

```python
next_tag = soup.select_one('a[rel="next"]')

if next_tag:
    next_url = urljoin(START_URL, next_tag["href"])
    print("다음 페이지:", next_url)
```

```python
def collect_all_books(start_url):
    collected = []
    url = start_url

    while url:
        response = requests.get(url, timeout=10)
        response.raise_for_status()
        soup = BeautifulSoup(response.text, "html.parser")

        page_rows = [
            parse_book(book, url)
            for book in soup.select("article.book")
        ]
        collected.extend(page_rows)
        print(url, "→", len(page_rows), "건")

        next_tag = soup.select_one('a[rel="next"]')
        url = urljoin(url, next_tag["href"]) if next_tag else None
        time.sleep(0.5)

    return collected

all_books = collect_all_books(START_URL)
print("전체 수집:", len(all_books), "건")
```
